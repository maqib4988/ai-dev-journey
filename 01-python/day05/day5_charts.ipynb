```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.read_csv("../day04/train.csv")
df["Age"] = df["Age"].fillna(df["Age"].median())

print(df.head())

plt.plot([1, 2, 3, 4, 5], [10, 25, 15, 30, 20])
plt.title("Simple Line Plot")
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.show()
```

```python
survival_by_class = df.groupby("Pclass")["Survived"].mean()
plt.bar(survival_by_class.index, survival_by_class.values)
plt.title("Survival Rate by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate")
plt.show()
```

```python
plt.hist(df["Age"], bins=20)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Count")
plt.show()
```

```python
plt.scatter(df["Age"], df["Fare"])
plt.title("Age vs Fare")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].bar(survival_by_class.index, survival_by_class.values)
axes[0].set_title("Survival by Class")
axes[1].hist(df["Age"], bins=20)
axes[1].set_title("Age Distribution")
plt.tight_layout()
plt.show()
```
